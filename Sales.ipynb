```python
import pandas as pd

sales = pd.read_csv("P20 source data/sales.csv")
```

```python
print(sales.columns)
```

```python
sales['order_date'] = pd.to_datetime(sales['order_date'])
sales['year'] = sales['order_date'].dt.year
```

```python
yearly = sales.groupby('year')['revenue'].sum()
print(yearly)
```

```python
yearly = yearly.reset_index()
print(yearly)
```

```python
yearly['growth %'] = yearly['revenue'].pct_change() * 100

print(yearly)
```
